```python
import numpy as np
from sklearn.preprocessing import PowerTransformer


def exp_b_log1p_zscore(vxs, CN05_PREs, precip_channels="all"):
    """
    Exp-B: log1p + z-score

    参数
    ----
    vxs : np.ndarray
        shape = (time, lat, lon, channel)
    CN05_PREs : np.ndarray
        shape = (time, lat, lon)
    precip_channels : "all" 或 list/int
        vxs 中需要先做 log1p 的降水相关通道索引
        例如 [0]、[0, 3]，或者 "all"

    返回
    ----
    vxs_out, CN05_PREs_out
    """
    vxs_out = np.array(vxs, dtype=np.float64, copy=True)
    CN05_PREs_out = np.array(CN05_PREs, dtype=np.float64, copy=True)

    # 1) 对 vxs 的降水相关通道做 log1p
    if precip_channels == "all":
        tmp = vxs_out
        if np.nanmin(tmp) < -1e-8:
            raise ValueError("vxs 中待做 log1p 的通道存在明显负值，不适合直接 log1p。")
        tmp[tmp < 0] = 0.0
        vxs_out = np.log1p(tmp)
    else:
        if isinstance(precip_channels, int):
            precip_channels = [precip_channels]
        tmp = vxs_out[..., precip_channels]
        if np.nanmin(tmp) < -1e-8:
            raise ValueError("vxs 中待做 log1p 的通道存在明显负值，不适合直接 log1p。")
        tmp[tmp < 0] = 0.0
        vxs_out[..., precip_channels] = np.log1p(tmp)

    # 2) 对目标 CN05_PREs 做 log1p
    if np.nanmin(CN05_PREs_out) < -1e-8:
        raise ValueError("CN05_PREs 存在明显负值，不适合直接 log1p。")
    CN05_PREs_out[CN05_PREs_out < 0] = 0.0
    CN05_PREs_out = np.log1p(CN05_PREs_out)

    # 3) 对整个 vxs 沿 time 维做 z-score
    vxs_mean = np.nanmean(vxs_out, axis=0, keepdims=True)
    vxs_std = np.nanstd(vxs_out, axis=0, keepdims=True)
    vxs_std[vxs_std == 0] = 1.0
    vxs_out = (vxs_out - vxs_mean) / vxs_std

    # 4) 对目标沿 time 维做 z-score
    y_mean = np.nanmean(CN05_PREs_out, axis=0, keepdims=True)
    y_std = np.nanstd(CN05_PREs_out, axis=0, keepdims=True)
    y_std[y_std == 0] = 1.0
    CN05_PREs_out = (CN05_PREs_out - y_mean) / y_std

    return vxs_out, CN05_PREs_out
```

```python
import numpy as np

def inverse_exp_b_predicty_testy(predicty, testy, CN05_PRE_ref):
    """
    Exp-B 反标准化:
    原流程: 原值 -> log1p -> z-score
    反过程: z-score -> log1p空间 -> expm1 -> 原值

    支持:
    - predicty/testy: (time, lat, lon) 或 (time, lat, lon, 1)
    - CN05_PRE_ref  : (time, lat, lon) 或 (time, lat, lon, 1)
    """

    def squeeze_last_if_one(arr):
        arr = np.array(arr, dtype=np.float64, copy=True)
        squeezed = False
        if arr.ndim == 4 and arr.shape[-1] == 1:
            arr = arr[..., 0]
            squeezed = True
        return arr, squeezed

    def add_last_if_needed(arr, squeezed):
        if squeezed:
            return arr[..., np.newaxis]
        return arr

    def safe_nanmean_std(arr):
        """
        arr: (time, lat, lon)
        沿 time 维计算 mean/std，避免 all-NaN warning
        """
        mask = np.isfinite(arr)
        count = np.sum(mask, axis=0, keepdims=True)

        arr0 = np.where(mask, arr, 0.0)
        mean = np.sum(arr0, axis=0, keepdims=True) / np.where(count == 0, 1, count)

        diff2 = np.where(mask, (arr - mean) ** 2, 0.0)
        var = np.sum(diff2, axis=0, keepdims=True) / np.where(count == 0, 1, count)
        std = np.sqrt(var)

        # 全 NaN 格点: mean=0, std=1
        mean = np.where(count == 0, 0.0, mean)
        std = np.where((count == 0) | (~np.isfinite(std)) | (std == 0), 1.0, std)

        return mean, std

    # -------------------------
    # 1. 统一成 3 维
    # -------------------------
    predicty_3d, predicty_was_4d = squeeze_last_if_one(predicty)
    testy_3d, testy_was_4d = squeeze_last_if_one(testy)
    ref_3d, _ = squeeze_last_if_one(CN05_PRE_ref)

    # -------------------------
    # 2. 参考值先做 log1p
    # -------------------------
    ref_3d[ref_3d < 0] = 0.0
    ref_log = np.log1p(ref_3d)

    # -------------------------
    # 3. 计算 z-score 参数
    # -------------------------
    y_mean, y_std = safe_nanmean_std(ref_log)   # (1, lat, lon)

    # -------------------------
    # 4. 先反 z-score
    # -------------------------
    predicty_log = predicty_3d * y_std + y_mean
    testy_log = testy_3d * y_std + y_mean

    # -------------------------
    # 5. 再反 log1p
    # -------------------------
    predicty_raw = np.expm1(predicty_log)
    testy_raw = np.expm1(testy_log)

    predicty_raw[predicty_raw < 0] = 0.0
    testy_raw[testy_raw < 0] = 0.0

    # -------------------------
    # 6. 恢复原始维度形式
    # -------------------------
    predicty_raw = add_last_if_needed(predicty_raw, predicty_was_4d)
    testy_raw = add_last_if_needed(testy_raw, testy_was_4d)

    return predicty_raw, testy_raw
predicty, testy = inverse_exp_b_predicty_testy(predicty, testy, CN05_PRE)
```
